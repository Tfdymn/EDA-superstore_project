<Now we will set Row ID as index to get make the dataset looks more organized

```python
df = df.set_index('Row ID')
df.tail()
```

<Now we will check the dtype of Order date and Ship date


```python
df[['Order Date', 'Ship Date']].dtypes
```

There datatype is string so we will convert it into Datetime

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df[['Order Date', 'Ship Date']].dtypes
```

```python
df.info()
```

**The dataset has no missing values or duplicate rows, and the date columns have been converted to datetime.**

```python
df.sample()
```

### Analyzing and Visualizing Data
**Let's check the total days taken to ship the order and add the Data in the Dataset**

```python
df['Shipping period'] = df['Ship Date'] - df['Order Date']
df.sample()
```

<Now let's check maximum days it took to ship any Order>

```python
df['Shipping period'].sort_values(ascending= False)
```

```python
sns.countplot(x= df['Shipping period'].sort_values(ascending= False), color= 'orange')
```

So the maximum days it took to ship any order is 7 days.

Most of the orders took 4 days to be shipped.
